## Contract

One synthetic visitor receives one randomized assignment and one simulated interface exposure. The binary outcome is form submission. The primary estimand is treatment minus control conversion probability. The +1.0 percentage-point practical threshold is a `PORTFOLIO_SCENARIO_ASSUMPTION`, not a company threshold.

## Interpretation and boundaries

The fixed seed and separate assignment/outcome streams are methodological controls, not a mechanism for selecting a favorable result.
